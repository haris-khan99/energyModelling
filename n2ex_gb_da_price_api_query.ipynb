```python
import requests
from urllib import parse

sql_query =  '''SELECT * FROM  "4f27eea5-7038-4f73-9740-e3e4ad47c26a" ORDER BY "_id" ASC LIMIT 100'''
params = {'sql': sql_query}

try:
    resposne = requests.get('https://api.neso.energy/api/3/action/datastore_search_sql', params = parse.urlencode(params))
    data = resposne.json()["result"]
    print(data) # Printing data
except requests.exceptions.RequestException as e:
    print(e.response.text)
```

```python
import requests
import pandas as pd

url = "https://api.neso.energy/api/3/action/datastore_search"

params = {
    "resource_id": "4f27eea5-7038-4f73-9740-e3e4ad47c26a",
    "limit": 1000,
    "offset": 0
}

response = requests.get(url, params=params, timeout=30)
response.raise_for_status()

data = response.json()
assert data["success"]

prices = pd.DataFrame(data["result"]["records"])
print(prices.head())
```

```python
import requests
import pandas as pd

url = (
    "https://data.elexon.co.uk/bmrs/api/v1/"
    "balancing/settlement/system-prices/2026-10-19"
)

response = requests.get(
    url,
    params={"format": "json"},
    timeout=30
)
response.raise_for_status()

data = response.json()

imbalance = pd.DataFrame(data["data"])
print(imbalance.tail(33))
```
